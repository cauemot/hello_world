# Automação de Sistemas e Processos com Python

### Desafio:

Todos os dias, o nosso sistema atualiza as vendas do dia anterior.
O seu trabalho diário, como analista, é enviar um e-mail para a diretoria, assim que começar a trabalhar, com o faturamento e a quantidade de produtos vendidos no dia anterior

E-mail da diretoria: seugmail+diretoria@gmail.com<br>
Local onde o sistema disponibiliza as vendas do dia anterior: https://drive.google.com/drive/folders/149xknr9JvrlEnhNWO49zPcw0PW5icxga

Para resolver isso, vamos usar o pyautogui, uma biblioteca de automação de comandos do mouse e do teclado


Referência do pyautogui: https://pyautogui.readthedocs.io/en/latest/quickstart.html

```python
# vamos precisar instalar
# pyautogui -> automações
# pandas -> dados
# openpyxl -> trabalhar com bases de dados em Excel (.xlsx)
# para instalar: abrir o terminal e rodar: pip install pandas pyautogui openpyxl

# pyautogui.click -> clicar com o mouse
# pyautogui.write -> escrever texto
# pyautogui.press -> pressionar tecla
# pyautogui.hotkey -> pressionar combinação de teclas
```

```python
import pyautogui
import time

# Passo a passo do desafio:
# Passo 1: Entrar no sistema da empresa (link do drive)
# Abrir o navegador padrão do sistema operacional
pyautogui.press('win')
pyautogui.write('opera') # Substitua pelo nome do navegador que você deseja abrir
pyautogui.press('enter')  

time.sleep(2)  # Aguardar o navegador abrir

# Digitei o link do drive
link = 'https://drive.google.com/drive/folders/149xknr9JvrlEnhNWO49zPcw0PW5icxga'
pyautogui.write(link)
# Apertei enter para acessar o link
pyautogui.press('enter')

time.sleep(4)  # Aguardar o drive abrir
```

```python
# Passo 2: Navegar no drive para encontar os dados da empresa
pyautogui.click(x=-1475, y=478, clicks=2)  # Clique em uma posição específica para focar na janela do navegador e da 2 clique para abrir a pasta "Exportar"
time.sleep(2)  # Aguardar a pasta abrir

# Passo 3: Baixar a base de dados
pyautogui.click(x=-1377, y=475, clicks=1)  # Clique nos 3 pontinhos no canto do arquivo "Vendas - Dez.xlsx"
time.sleep(1)  # Aguardar o menu abrir
pyautogui.click(x=-1321, y=565, clicks=1)  # Clique na opção baixar
time.sleep(3)  # Aguardar o download completar
```

### Vamos agora ler o arquivo baixado para pegar os indicadores

- Faturamento
- Quantidade de Produtos

```python
import pandas

# Passo 4: Importar a base de dados para o Python
# Abrir a base de dados com o pandas
caminho = r"C:\Users\kawem\Downloads\Vendas - Dez.xlsx"
tabela = pandas.read_excel(caminho)

# Ver as informações da base de dados
display(tabela)

# Somar o faturamento total dos produtos vendidos = somar a coluna "Valor Final"
faturamento = tabela["Valor Final"].sum()

# Somar a quantidade total de produtos vendidos = somar a coluna "Quantidade"
qtde_produtos = tabela["Quantidade"].sum()

print(faturamento)
print(qtde_produtos)
```

### Vamos agora enviar um e-mail pelo gmail

```python
# Passo 5: Enviar um e-mail pelo Gmail para o diretor da empresa com os resultados (faturamento e quantidade de produtos vendidos)
import pyperclip


pyautogui.hotkey('ctrl', 't')  # Abrir uma nova aba no navegador

pyautogui.write('https://mail.google.com/')  # Digitar o link do Gmail
pyautogui.press('enter')  # Pressionar Enter para acessar o Gmail
time.sleep(3)  # Aguardar o Gmail abrir

# Clicar no botão "Escrever" para iniciar um novo e-mail
pyautogui.click(x=-1772, y=211) 
time.sleep(2)  # Aguardar a janela de composição do e-mail abrir

# Digitar o destinatário do e-mail
pyautogui.write('xxx@gmail.com')  # Substitua pelo e-mail do diretor da empresa
pyautogui.press('tab')  # Pressionar Tab para ir para o campo do assunto

# Assunto do email
pyperclip.copy('Relatório de Vendas')  # Substitua pelo assunto do e-mail
pyautogui.hotkey('ctrl', 'v')  # Colar o assunto do e-mail
pyautogui.press('tab')  # Pressionar Tab para ir para o corpo do e-mail

# Qual o corpo do email
texto = f"""
Prezados, Diretores

Segue o relatório de vendas do mês.

Faturamento total: R${faturamento:,.2f}
Quantidade total de produtos vendidos: {qtde_produtos:,}

Atenciosamente,
Cauê
"""
pyperclip.copy(texto)  # Copiar o corpo do e-mail
pyautogui.hotkey('ctrl', 'v')  # Colar o corpo do e-mail

# Enviar o email
pyautogui.click(x=-624, y=995) # Clique no botão "Enviar"

```
