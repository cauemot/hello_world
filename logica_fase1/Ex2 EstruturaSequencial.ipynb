#### 11. Tendo como dado de entrada a altura (h) de uma pessoa, construa um algoritmo que calcule seu peso ideal, utilizando as seguintes fórmulas:

#### 12. Faça um Programa que pergunte quanto você ganha por hora e o número de horas trabalhadas no mês.

```python
## Descobrir quanto ganha por hora
din_hora = float(input("Quanto voce ganha por hora? "))
## Decobrir quantas horas trabalha por mes
horas_mes = float(input("Quantas horas voce trabalha por mes? "))
```

#####  Calcule o salário bruto (horas * salario por hora)

```python
## Calcular o salario bruto
salario_bruto = din_hora * horas_mes
print(f"Seu salario bruto e de R${salario_bruto:,.2f}.")
```

##### Calcule o desconto do IR (11% do salário bruto)

```python
## Cacular o desconto do Imposto de Renda (11% do salario bruto)
ir = salario_bruto * 0.11
print(f"O desconto do imposto de renda e de R${ir:,.2f}.")
```

##### Calcule o desconto do INSS (8% do salário bruto)

```python
## Calcular o desconto do INSS (8% do salario bruto)
inss = salario_bruto * 0.08
print(f"O desconto do INSS e de R${inss:,.2f}.")
```

##### Calcule o desconto do sindicato (5% do salário bruto)

```python
## Calcular o desconto do sindicato (5% do salario bruto)
sindicato = salario_bruto * 0.05
print(f"O desconto do sindicato e de R${sindicato:,.2f}.")
```

##### Calcule o salário líquido (salário bruto - descontos)

```python
## Calcular o salario liquido
descontos = ir + inss + sindicato
salario_liquido = salario_bruto - descontos
print(f"Seu salario liquido e de R${salario_liquido:,.2f}.")
```

#### 13. Faça um programa para uma loja de tintas. O programa deverá pedir o tamanho em metros quadrados da área a ser pintada. Considere que a cobertura da tinta é de 1 litro para cada 3 metros quadrados e que a tinta é vendida em latas de 18 litros, que custam R\$ 80,00. Informe ao usuário a quantidades de latas de tinta a serem compradas e o preço total. (para simplificação nesse momento, não se preocupe em arredondar a quantidade de latas a serem compradas - vamos trabalhar isso em breve)

```python
largura = float(input("Digite a largura da parede em metros: "))
altura = float(input("Digite a altura da parede em metros: "))
area = largura * altura 
print(f"A area da parede e de {area:.2f} metros quadrados.")

litro_lata = 18
cobertura_lata = 3
qtde_latas = area / (litro_lata * cobertura_lata)
print(f"A quantidade de latas de tinta a serem compradas e de {qtde_latas:.2f} latas.")

preco_lata = 80
preco_total = qtde_latas * preco_lata
print(f"O valor total a ser pago pelas latas de tinta e de R$ {preco_total:.2f}.")
```

#### 14. Faça um programa que peça o tamanho de um arquivo para download (em MB) e a velocidade de um link de Internet (em Mbps), calcule e informe o tempo aproximado de download do arquivo usando este link (em minutos).

Detalhe: MB significa megabyte, Mb (com b minúsculo) significa megabit. Um megabit é 1/8 de um megabyte. 

```python
try:
    mb = float(input("Digite o tamanho do arquivo em MB: "))
    mbps = float(input("Digite a velocidade da internet em Mbps: "))

    if mb <= 0 or mbps <= 0:
        print("Erro: O tamanho do arquivo e a velocidade da internet devem ser maiores que zero.")
    else:
        tempo = (mb * 8) / mbps
        minutos = tempo / 60
        print(f"O tempo aproximado de download do arquivo e de {minutos:.2f} minutos.")
except ValueError:
    print("Erro: Por favor, digite valores numéricos válidos.")
```
